```python
# create know parametes
weight = 0.7
bias = 0.3

#create
start = 0
end = 1
step = 0.02
X = torch.arange(start, end, step).unsqueeze(dim=1)
y = weight * X + bias
print(X[:10], y[:10])
```

```python
 train_split = int(0.8 * len(X))
 X_train, y_train = X[:train_split], y[:train_split]
 X_test, y_test = X[train_split:], y[train_split:]
 len(X_train), len(y_train), len(X_test), len(y_test)

 # in further we will use sklearn train, validate, test function to divide it, it have some more cool, property which are helpful
```

* now we will create a function to visulise the input and output data, so that we can also use that in future


```python
import matplotlib.pyplot as plt
def plot_predictions(train_data=X_train,
                     train_labels=y_train,
                     test_data=X_test,
                     test_labels=y_test,
                     predictions=None):
  plt.figure(figsize=(10, 7))
  # Plot training data in blue
  plt.scatter(train_data, train_labels, c="b", s=4, label="Training data")
  # plot testing data in green
  plt.scatter(test_data, test_labels, c="g", s=4, label="Testing data")
  if predictions is not None:
    # plot the predictions in red (predictions were made on the test data)
    plt.scatter(test_data, predictions, c="r", s=4, label="Predictions")
  plt.legend(prop={'size': 14})
  plt.show()

plot_predictions(X_train, y_train, X_test, y_test)
```

* lets make our first pytorch nn

```python
# we will make a linear regression model.
class LinearRegressionModelV1(nn.Module): # almost everything in pytorch neural netwroks, inherits from nn.module so your class should have nn.module as a sub class
  def __init__(self):
    super().__init__() # dk what this is
    self.weights = nn.Parameter(torch.randn(1, requires_grad=True, dtype=torch.float))# randn  gives a random value with a normal distribution mean =0 and variance is 1
    self.bias = nn.Parameter(torch.randn(1, requires_grad=True, dtype=torch.float))
  def forward(self, x: torch.Tensor) -> torch.Tensor: # x is the input data meaning is that x is a torch.tensor format input so is the output.
    return self.weights * x + self.bias
```

### pytorch, model beginning essetials
torch.nn = contains all of the buildings for
 1. computational graphs (a neural network can be considered a computational graph

 2. torch.nn.parameter- what parameters should our model try and learn, often a pytorch layer from torch.nn will set this for us

 3. torch.nn.module - the base class for all neural networks modules, if you subclass it, you should overwrite forward()

 4. torch.optim  - this is done to optimize our model, like Adam and all

```python
 # as above we can see this is shit, so now we will, optimize it. but lets first train our model by getting loss function
loss_fn = nn.L1Loss()

# setting a optimizer
optimizer = torch.optim.SGD(params = model_0.parameters(), lr=0.01 ) #no momentum 😭😭
```

```python
# building a training loop yoo lets go an epoch is one loop thorugh the process of data to optimizing
epochs = 1000
for epoch in range(epochs):
  # set the model to training mode
  model_0.train() # train mode in putorch sets all parameters that require gradients

  #forward pass
  y_pred = model_0(X_train)

  #loss
  loss = loss_fn(y_pred, y_train)

  #optimizer zero grad
  optimizer.zero_grad() # make the optimizer go fresh each time

  # perform backprop
  loss.backward()

  #step the optimizer
  optimizer.step()

  #testing

  model_0.eval() # turns of gradient tracking and useless tracking
  with torch.no_grad():
    test_pred = model_0(X_test)
    # loss
    test_loss = loss_fn(test_pred, y_test)
  if epoch%100 == 0:
    print(f'Epoch: {epoch} | Loss: {loss} | Test_loss: {test_loss}')


print(model_0.state_dict())
 #in just 100 loops our loss is already around 0.02, and values quite close to true weight and bias , which is good
```

```python
# now lets go through all the previous step to summarize

#first of all we import all the necessay thing
import torch
from torch import nn
import matplotlib.pyplot as plt


# chekcking pytorch version so if it is old some code might not work
torch.__version__
```

```python
# now lets do some device aganostics meanign if GPU is available out code will use that for faster calculation

device = 'cuda' if torch.cuda.is_available() else 'cpu'
print('device using to code is', device)
```

```python
# now lets make some data to train our model on
weight = 0.3
bias = 0.7

start = 0
end = 1
step=0.02

X = torch.arange(start, end, step).unsqueeze(dim=1) # mentioning unsqueeze dimension is impotant
Y = weight*X + bias
print(Y)
# now lets add some noise in it, to make it little random
total_data_points = int((end-start)/step)
print(total_data_points)
noise = torch.rand(total_data_points).unsqueeze(axis=1)
noise = noise*2-1
noise = noise/100 # reducing the noise amplitude,
Y = Y + noise # adding noise
print(Y)
```

```python
plt.plot(X,Y)
plt.show()
```

```python
# now, lets make training and test datasets
train_split = int(0.8*len(X))# so its a 80 20 split
X_train,Y_train = X[: train_split],Y[:train_split]
X_test, Y_test = X[train_split:], Y[train_split:]
len(X_train), len(Y_train), len(X_test), len(Y_test)
```

```python
import matplotlib.pyplot as plt
def plot_predictions(train_data=X_train,
                     train_labels=Y_train,
                     test_data=X_test,
                     test_labels=Y_test,
                     predictions=None):
  plt.figure(figsize=(10, 7))
  # Plot training data in blue
  plt.scatter(train_data, train_labels, c="b", s=4, label="Training data")
  # plot testing data in green
  plt.scatter(test_data, test_labels, c="g", s=4, label="Testing data")
  if predictions is not None:
    # plot the predictions in red (predictions were made on the test data)
    plt.scatter(test_data, predictions, c="r", s=4, label="Predictions")
  plt.legend(prop={'size': 14})
  plt.show()

plot_predictions(X_train, Y_train, X_test, Y_test)
```

```python
# now lets create a linear model by subclassing nn.module
class LinearRegressionModelV2(nn.Module):
  def __init__(self):
    super().__init__()
    # use nn.Linear() for creating the model parameters initial we initialized params ourslef, but we can do it this way also which is more effective if there are many layers of parameters
    self.Linear_layer = nn.Linear(in_features=1 , out_features=1)
  def forward(self, X: torch.Tensor)->torch.Tensor:
    return self.Linear_layer(X)
# set the seed for reproduclibility
torch.manual_seed(42)
model_1 = LinearRegressionModelV2()# defines model
model_1, model_1.state_dict()# print the weights and baises
```

```python
# now we will make sure and see once where all the computation will happen in cpu or gpu,
# so currently what the model is on we can check by
next(model_1.parameters()).device
```

```python
# now if we want to set it to cuda gpu , we do
model_1.to(device)
next(model_1.parameters()).device
# now our parameters are in cuda gpu, so now our computation will happen in cuda
```

```python
# now lets train our model
 # lets first train our model by getting loss function
loss_fn = nn.L1Loss()

# setting a optimizer
optimizer = torch.optim.SGD(params = model_1.parameters(), lr=0.01 ) #no momentum 😭😭
```

```python
# lets write manual seed and build a training epoch
torch.manual_seed(42)
epochs = 200
X_train = X_train.to(device) # this is important customization so dont forset to write this
Y_train = Y_train.to(device)
X_test = X_test.to(device)
Y_test = Y_test.to(device)


# putting our data on gpu or like simply putting device agnostic code

for epoch in range(epochs):
  # putting our model in training mode
  model_1.train() # train mode in putorch sets all parameters that require gradients

  #forward pass
  Y_pred = model_1(X_train) # this code goes thorugh the model _1 and implement forward return here
  # this will give error because our model is in cpu but our data is on cpu so we will also have to put our data on gpu
  #loss
  loss = loss_fn(Y_pred, Y_train)

  #optimizer zero grad
  optimizer.zero_grad() # make the optimizer go fresh each time

  # perform backprop
  loss.backward()

  #step the optimizer
  optimizer.step()

  #testing

  model_1.eval() # turns of gradient tracking and useless tracking
  with torch.no_grad():
    test_pred = model_1(X_test)
    # loss
    test_loss = loss_fn(test_pred, Y_test)
  if epoch%10 == 0:
    print(f'Epoch: {epoch} | Loss: {loss} | Test_loss: {test_loss}')


print(model_1.state_dict())
 #in just 100 loops our los
```

```python
# now we will evaluate prediction but first we will go on eval mode
model_1.eval()

with torch.no_grad():# using this make the gradient tracking off to  make the model fast, for inference use this if infernece is not working as above
  Y_pred = model_1(X_test) # just doing so make the forward in the nn run, and return the output as y_pred, this is good, but what happen if there are backwards functions and other things also?
  plot_predictions(X_train.cpu(), Y_train.cpu(), X_test.cpu(), Y_test.cpu(), Y_pred.cpu())
# converting them to cpu is necessay because matplotlib is not accesible from gpu, so .cpu() converted them to cpu first
# this just convert the like X_train and other to cpu for plotting but the all the values are only gpu only
print(Y_pred)
```
